```python
from result import ExperimentResult
import numpy as np
import seaborn as sns
experiment_name = "6_rastrigin_sigma_N_lamda_small_lamda"
experiment_result = ExperimentResult.from_csv(f"../results/{experiment_name}.csv")
df = experiment_result.to_dataframe()

# Calculate success probability
df["Success Probability"] = df["Success"].apply(np.mean)

sns.set_theme(style="whitegrid")
# get the parameters from the configuration
d = df.iloc[0].loc["d"]
max_it = df.iloc[0].loc["max_it"]
alpha = df.iloc[0].loc["alpha"]
dt = df.iloc[0].loc["dt"]
f = df.iloc[0].loc["Function Name"]

num_sigmas = len(df["sigma"].unique())
num_N = len(df["N"].unique())

dynamic_name = "qcbo"
df_dynamic = df[df["Dynamic Name"] == dynamic_name]
dfm = df_dynamic.melt(
        id_vars=["Dynamic Name", "sigma", "Function Name", "N", "lamda"],
        value_vars=["Success Probability"],
        var_name="variable",
        value_name="success_prob",
    )#.pivot(index="N", columns="sigma", values="success_prob")

import plotly.express as px
# Create the interactive heatmap
fig = px.density_heatmap(
    dfm,
    x='sigma',
    y='N',
    z='success_prob',
    nbinsx=num_sigmas, nbinsy=num_N,
    animation_frame='lamda',  # This creates a slider for param3
    color_continuous_scale='Viridis',
    title=f"Success Probability for {dynamic_name} and {f} at d={d}, max_it={max_it}, alpha={alpha}, dt={dt}"
)
# fig.set_yticklabels(fig.get_yticklabels(), rotation=45)
# fig.invert_yaxis()

# Display the plot
fig.update_layout(
    xaxis_title="Sigma",
    yaxis_title="Number of Particles",
    coloraxis_colorbar=dict(title="Success Probability")
)

fig.write_html(f"../results/html/sigma_N_{experiment_name}_{dynamic_name}.html")
```

```python
from result import ExperimentResult
import numpy as np
import seaborn as sns
experiment_name = "6_rastrigin_sigma_n_lamda_cbo"
experiment_result = ExperimentResult.from_csv(f"../results/{experiment_name}.csv")
df = experiment_result.to_dataframe()

# Calculate success probability
df["Success Probability"] = df["Success"].apply(np.mean)

sns.set_theme(style="whitegrid")
# get the parameters from the configuration
d = df.iloc[0].loc["d"]
max_it = df.iloc[0].loc["max_it"]
alpha = df.iloc[0].loc["alpha"]
dt = df.iloc[0].loc["dt"]
f = df.iloc[0].loc["Function Name"]

num_sigmas = len(df["sigma"].unique())
num_lamdas = len(df["lamda"].unique())

dynamic_name = "cbo"
df_dynamic = df[df["Dynamic Name"] == dynamic_name]
dfm = df_dynamic.melt(
        id_vars=["Dynamic Name", "sigma", "Function Name", "N", "lamda"],
        value_vars=["Success Probability"],
        var_name="variable",
        value_name="success_prob",
    )#.pivot(index="N", columns="sigma", values="success_prob")

import plotly.express as px

# Calculate the global min and max for success_prob
min_val = dfm['success_prob'].min()
max_val = dfm['success_prob'].max()
# Create the interactive heatmap
fig = px.density_heatmap(
    dfm,
    x='sigma',
    y='lamda',
    z='success_prob',
    nbinsx=num_sigmas, nbinsy=num_lamdas,
    animation_frame='N',  # This creates a slider for param3
    color_continuous_scale='Viridis',
    range_color=[min_val, max_val],  # Set the fixed range for the color scale
    title=f"Success Probability for {dynamic_name} and {f} at d={d}, max_it={max_it}, alpha={alpha}, dt={dt}"
)
# fig.set_yticklabels(fig.get_yticklabels(), rotation=45)
# fig.invert_yaxis()

# Display the plot
fig.update_layout(
    xaxis_title="Sigma",
    yaxis_title="Lamda",
    coloraxis_colorbar=dict(title="Success Probability")
)

fig.write_html(f"../results/html/sigma_lamda_{experiment_name}_{dynamic_name}.html")
```
