```python
import cbx
import numpy as np
def execute_dynamic(config, starting_positions):
    opt_dict = {"sched": None, "print_int": None}
    func = cbx.objectives.Rastrigin()
    dynamic = cbx.dynamics.CBO(func, x=starting_positions, **config)
    best_x = dynamic.optimize(**opt_dict)
    best_f = func(best_x)
    success = best_f < 1
    return best_x, best_f, success

cfg_base = {'f_dim': '3D', 'check_f_dims': True, 'x_min': 0.5, 'x_max': 5.0, 'M': 1000, 'N': 10, 'd': 4, 'max_it': 500, 'term_criteria': None, 'track_args': None, 'verbosity': 0, 'copy': None, 'norm': None, 'normal': None, 'post_process': None, 'noise': 'anisotropic', 'batch_args': None, 'dt': None, 'alpha': 1, 'sigma': 1.0, 'lamda': 1.0, 'correction': 'no_correction', 'correction_eps': 0.001, 'compute_consensus': None}
starting_positions = cbx.particle_init.init_particles(
                    shape=(cfg_base["M"], cfg_base["N"], cfg_base["d"]), 
                    x_min = -1.0, x_max = 1.0
                )

# test with 0.1 and 0.02
cfg_1 = cfg_base.copy()
cfg_1["dt"] = 0.1
result_1 = execute_dynamic(cfg_1, starting_positions)

cfg_2 = cfg_base.copy()
cfg_2["dt"] = 0.02
cfg_2["sigma"] = 2.23606 # sqrt 5
cfg_2["lamda"] = 5.0
result_2 = execute_dynamic(cfg_2, starting_positions)

print("Configurations")
print(f"Cfg 1: dt={cfg_1['dt']}, lamda={cfg_1['lamda']}, sigma={cfg_1['sigma']}")
print(f"Cfg 2: dt={cfg_2['dt']}, lamda={cfg_2['lamda']}, sigma={cfg_2['sigma']}")
print()
print("Results - Mean Fitness")
print(f"Cfg 1: {np.mean(result_1[1]).round(2)}")
print(f"Cfg 2: {np.mean(result_2[1]).round(2)}")
print()
print("Results - Success")
print(f"Cfg 1: {np.mean(result_1[2]).round(2)}")
print(f"Cfg 2: {np.mean(result_2[2]).round(2)}")
print()
print("Base Config")
print(cfg_base)
```
